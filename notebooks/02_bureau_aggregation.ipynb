Aggregation

```python
# ============================================================
# 1. MISSINGNESS FLAGS
# ============================================================

df['HAS_MAX_OVERDUE_RECORD'] = (
    df['AMT_CREDIT_MAX_OVERDUE'].notna().astype(int)
)

df['HAS_ANNUITY'] = (
    df['AMT_ANNUITY'].notna().astype(int)
)


# ============================================================
# 2. CLEAN CREDIT TYPE
# ============================================================

type_mapping = {
    'Consumer credit': 'Consumer',
    'Credit card': 'Credit_Card',
    'Car loan': 'Car_Mortgage',
    'Mortgage': 'Car_Mortgage'
}

df['CREDIT_TYPE_CLEAN'] = (
    df['CREDIT_TYPE']
    .map(type_mapping)
    .fillna('Other')
)


# ============================================================
# 3. OVERALL CUSTOMER-LEVEL AGGREGATION
# ============================================================

agg_rules = {

    # Number of previous bureau credit records
    'SK_ID_BUREAU': 'count',

    # Credit history timing
    'DAYS_CREDIT': ['min', 'max', 'mean'],

    # Overdue behaviour
    'CREDIT_DAY_OVERDUE': ['max', 'mean'],

    # Credit amounts
    'AMT_CREDIT_SUM': ['sum', 'mean', 'max'],

    # Debt
    'AMT_CREDIT_SUM_DEBT': ['sum', 'mean', 'max'],

    # Overdue amount
    'AMT_CREDIT_SUM_OVERDUE': ['sum', 'max'],

    # Credit prolongations
    'CNT_CREDIT_PROLONG': ['sum', 'max'],

    # Missing-information indicators
    'HAS_MAX_OVERDUE_RECORD': 'sum',
    'HAS_ANNUITY': 'sum'
}


df_all = (
    df
    .groupby('SK_ID_CURR')
    .agg(agg_rules)
)


# Rename aggregated columns
df_all.columns = [
    f'BUREAU_{col}_{agg.upper()}'
    for col, agg in df_all.columns
]


# Make names more business-friendly
df_all = df_all.rename(columns={

    'BUREAU_SK_ID_BUREAU_COUNT':
        'BUREAU_TOTAL_LOANS',

    'BUREAU_DAYS_CREDIT_MIN':
        'BUREAU_DAYS_CREDIT_MIN',

    'BUREAU_DAYS_CREDIT_MAX':
        'BUREAU_DAYS_CREDIT_MAX',

    'BUREAU_DAYS_CREDIT_MEAN':
        'BUREAU_DAYS_CREDIT_MEAN',

    'BUREAU_CREDIT_DAY_OVERDUE_MAX':
        'BUREAU_OVERDUE_DAYS_MAX',

    'BUREAU_CREDIT_DAY_OVERDUE_MEAN':
        'BUREAU_OVERDUE_DAYS_MEAN',

    'BUREAU_AMT_CREDIT_SUM_SUM':
        'BUREAU_TOTAL_CREDIT',

    'BUREAU_AMT_CREDIT_SUM_MEAN':
        'BUREAU_AVG_CREDIT',

    'BUREAU_AMT_CREDIT_SUM_MAX':
        'BUREAU_MAX_CREDIT',

    'BUREAU_AMT_CREDIT_SUM_DEBT_SUM':
        'BUREAU_TOTAL_DEBT',

    'BUREAU_AMT_CREDIT_SUM_DEBT_MEAN':
        'BUREAU_AVG_DEBT',

    'BUREAU_AMT_CREDIT_SUM_DEBT_MAX':
        'BUREAU_MAX_DEBT',

    'BUREAU_AMT_CREDIT_SUM_OVERDUE_SUM':
        'BUREAU_TOTAL_OVERDUE',

    'BUREAU_AMT_CREDIT_SUM_OVERDUE_MAX':
        'BUREAU_MAX_OVERDUE',

    'BUREAU_CNT_CREDIT_PROLONG_SUM':
        'BUREAU_TOTAL_CREDIT_PROLONG',

    'BUREAU_CNT_CREDIT_PROLONG_MAX':
        'BUREAU_MAX_CREDIT_PROLONG',

    'BUREAU_HAS_MAX_OVERDUE_RECORD_SUM':
        'BUREAU_MAX_OVERDUE_RECORD_COUNT',

    'BUREAU_HAS_ANNUITY_SUM':
        'BUREAU_ANNUITY_RECORD_COUNT'
})


# ============================================================
# 4. ACTIVE CREDIT AGGREGATION
# ============================================================

df_active = (
    df[df['CREDIT_ACTIVE'] == 'Active']
    .groupby('SK_ID_CURR')
    .agg({
        'SK_ID_BUREAU': 'count',
        'AMT_CREDIT_SUM': ['sum', 'mean'],
        'AMT_CREDIT_SUM_DEBT': ['sum', 'mean'],
        'CREDIT_DAY_OVERDUE': ['max', 'mean']
    })
)


df_active.columns = [
    f'BUREAU_ACTIVE_{col}_{agg.upper()}'
    for col, agg in df_active.columns
]


# Rename active features
df_active = df_active.rename(columns={

    'BUREAU_ACTIVE_SK_ID_BUREAU_COUNT':
        'BUREAU_ACTIVE_LOANS',

    'BUREAU_ACTIVE_AMT_CREDIT_SUM_SUM':
        'BUREAU_ACTIVE_TOTAL_CREDIT',

    'BUREAU_ACTIVE_AMT_CREDIT_SUM_MEAN':
        'BUREAU_ACTIVE_AVG_CREDIT',

    'BUREAU_ACTIVE_AMT_CREDIT_SUM_DEBT_SUM':
        'BUREAU_ACTIVE_TOTAL_DEBT',

    'BUREAU_ACTIVE_AMT_CREDIT_SUM_DEBT_MEAN':
        'BUREAU_ACTIVE_AVG_DEBT',

    'BUREAU_ACTIVE_CREDIT_DAY_OVERDUE_MAX':
        'BUREAU_ACTIVE_OVERDUE_DAYS_MAX',

    'BUREAU_ACTIVE_CREDIT_DAY_OVERDUE_MEAN':
        'BUREAU_ACTIVE_OVERDUE_DAYS_MEAN'
})


# ============================================================
# 5. CREDIT TYPE COUNTS
# ============================================================

credit_type_counts = pd.crosstab(
    df['SK_ID_CURR'],
    df['CREDIT_TYPE_CLEAN']
)


credit_type_counts.columns = [
    f'BUREAU_CREDIT_TYPE_{col.upper()}_COUNT'
    for col in credit_type_counts.columns
]


# ============================================================
# 6. COMBINE ALL BUREAU FEATURES
# ============================================================

final_bureau_features = (
    df_all
    .join(df_active, how='left')
    .join(credit_type_counts, how='left')
)


# Customers without active loans/type records → 0
final_bureau_features = final_bureau_features.fillna(0)


# ============================================================
# 7. RESET INDEX
# ============================================================

final_bureau_features = (
    final_bureau_features
    .reset_index()
)
```

```python
# ============================================================
# 8. FINAL CHECK
# ============================================================

print("Shape:", final_bureau_features.shape)

print(
    "\nUnique customers:",
    final_bureau_features['SK_ID_CURR'].nunique()
)

print(
    "Duplicate customer IDs:",
    final_bureau_features['SK_ID_CURR'].duplicated().sum()
)

print("\nSample:")
display(final_bureau_features.head())
```
