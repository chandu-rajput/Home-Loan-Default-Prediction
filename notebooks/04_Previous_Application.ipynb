## **Understand all columns**


| Column | Decision | Reason |
| :--- | :--- | :--- |
| SK_ID_PREV | Use | Count/mapping ID, not a predictive value |
| SK_ID_CURR | Grouping key | Customer ID |
| NAME_CONTRACT_STATUS | Definitely use | Approved/refused history |
| NAME_CONTRACT_TYPE | Use | Previous loan types |
| AMT_APPLICATION | Use | Requested amount |
| AMT_CREDIT | Use | Granted credit |
| AMT_ANNUITY | Use | Payment burden |
| AMT_DOWN_PAYMENT | Use | Financing structure |
| CNT_PAYMENT | Use | Repayment term |
| DAYS_DECISION | Use | Recency of previous application |
| NAME_CLIENT_TYPE | Use | New/repeater customer behavior |
| NAME_PORTFOLIO | Use | Previous product portfolio |
| NAME_PRODUCT_TYPE | Use | Product/application type |
| CHANNEL_TYPE | Maybe use | Application channel |
| CODE_REJECT_REASON | Maybe use | Historical rejection reason |
| NAME_CASH_LOAN_PURPOSE | Low priority | Lots of XNA/XAP |
| NAME_PAYMENT_TYPE | Low priority | Less directly useful |
| NAME_TYPE_SUITE | Low priority | High missingness |
| RATE_INTEREST_PRIMARY | Skip initially | 99.64% missing |
| RATE_INTEREST_PRIVILEGED | Skip initially | 99.64% missing |
| DAYS_FIRST_DRAWING etc. | Skip initially | Missing ~40%, needs more interpretation |
| WEEKDAY_APPR_PROCESS_START | Skip initially | Little customer-level value |
| HOUR_APPR_PROCESS_START | Skip initially | Weak business relevance |


Aggregation

```python
# ============================================================
# 1. NUMERICAL / TIME-BASED AGGREGATION
# ============================================================

agg_rules = {

    # Number of previous applications
    'SK_ID_PREV': 'count',

    # Requested application amount
    'AMT_APPLICATION': ['sum', 'mean', 'max'],

    # Credit amount granted
    'AMT_CREDIT': ['sum', 'mean', 'max'],

    # Loan annuity / payment amount
    'AMT_ANNUITY': ['mean', 'max'],

    # Down payment
    'AMT_DOWN_PAYMENT': ['mean', 'max'],

    # Goods price
    'AMT_GOODS_PRICE': ['mean', 'max'],

    # Number of payments / loan term
    'CNT_PAYMENT': ['mean', 'max'],

    # Down payment rate
    'RATE_DOWN_PAYMENT': ['mean', 'max'],

    # How recently previous applications were decided
    'DAYS_DECISION': ['min', 'max', 'mean']
}


prev_all = (
    df
    .groupby('SK_ID_CURR')
    .agg(agg_rules)
)


prev_all.columns = [
    f'PREV_APP_{col}_{agg.upper()}'
    for col, agg in prev_all.columns
]
```

```python
# ============================================================
# 2. BUSINESS-FRIENDLY FEATURE NAMES
# ============================================================

prev_all = prev_all.rename(columns={

    'PREV_APP_SK_ID_PREV_COUNT':
        'PREV_APP_TOTAL_APPLICATIONS',

    'PREV_APP_AMT_APPLICATION_SUM':
        'PREV_APP_TOTAL_APPLICATION_AMOUNT',

    'PREV_APP_AMT_APPLICATION_MEAN':
        'PREV_APP_AVG_APPLICATION_AMOUNT',

    'PREV_APP_AMT_APPLICATION_MAX':
        'PREV_APP_MAX_APPLICATION_AMOUNT',

    'PREV_APP_AMT_CREDIT_SUM':
        'PREV_APP_TOTAL_CREDIT',

    'PREV_APP_AMT_CREDIT_MEAN':
        'PREV_APP_AVG_CREDIT',

    'PREV_APP_AMT_CREDIT_MAX':
        'PREV_APP_MAX_CREDIT',

    'PREV_APP_AMT_ANNUITY_MEAN':
        'PREV_APP_AVG_ANNUITY',

    'PREV_APP_AMT_ANNUITY_MAX':
        'PREV_APP_MAX_ANNUITY',

    'PREV_APP_AMT_DOWN_PAYMENT_MEAN':
        'PREV_APP_AVG_DOWN_PAYMENT',

    'PREV_APP_AMT_DOWN_PAYMENT_MAX':
        'PREV_APP_MAX_DOWN_PAYMENT',

    'PREV_APP_AMT_GOODS_PRICE_MEAN':
        'PREV_APP_AVG_GOODS_PRICE',

    'PREV_APP_AMT_GOODS_PRICE_MAX':
        'PREV_APP_MAX_GOODS_PRICE',

    'PREV_APP_CNT_PAYMENT_MEAN':
        'PREV_APP_AVG_PAYMENT_COUNT',

    'PREV_APP_CNT_PAYMENT_MAX':
        'PREV_APP_MAX_PAYMENT_COUNT',

    'PREV_APP_RATE_DOWN_PAYMENT_MEAN':
        'PREV_APP_AVG_DOWN_PAYMENT_RATE',

    'PREV_APP_RATE_DOWN_PAYMENT_MAX':
        'PREV_APP_MAX_DOWN_PAYMENT_RATE',

    'PREV_APP_DAYS_DECISION_MIN':
        'PREV_APP_OLDEST_DECISION',

    'PREV_APP_DAYS_DECISION_MAX':
        'PREV_APP_MOST_RECENT_DECISION',

    'PREV_APP_DAYS_DECISION_MEAN':
        'PREV_APP_AVG_DECISION_TIME'
})
```

```python
# ============================================================
# 3. CONTRACT STATUS COUNTS
# ============================================================

status_counts = pd.crosstab(
    df['SK_ID_CURR'],
    df['NAME_CONTRACT_STATUS']
)

status_counts.columns = [
    f'PREV_APP_STATUS_{col.upper().replace(" ", "_")}_COUNT'
    for col in status_counts.columns
]

# Total applications
status_counts['PREV_APP_TOTAL_APPLICATIONS'] = (
    status_counts.sum(axis=1)
)

# Approval rate
status_counts['PREV_APP_APPROVAL_RATE'] = (
    status_counts['PREV_APP_STATUS_APPROVED_COUNT']
    / status_counts['PREV_APP_TOTAL_APPLICATIONS'].replace(0, np.nan)
)

# Refusal rate
status_counts['PREV_APP_REFUSAL_RATE'] = (
    status_counts['PREV_APP_STATUS_REFUSED_COUNT']
    / status_counts['PREV_APP_TOTAL_APPLICATIONS'].replace(0, np.nan)
)

status_counts.drop(columns='PREV_APP_TOTAL_APPLICATIONS',axis=1, inplace=True)


status_counts.head(2)
```

```python
# ============================================================
# 4. CONTRACT TYPE COUNTS
# ============================================================

contract_type_counts = pd.crosstab(
    df['SK_ID_CURR'],
    df['NAME_CONTRACT_TYPE']
)

contract_type_counts.columns = [
    f'PREV_APP_CONTRACT_{col.upper().replace(" ", "_")}_COUNT'
    for col in contract_type_counts.columns
]

contract_type_counts = contract_type_counts.drop(
    columns=['PREV_APP_CONTRACT_XNA_COUNT'],
    errors='ignore'
)

contract_type_counts.head(2)
```

```python
# ============================================================
# 5. CLIENT TYPE COUNTS
# ============================================================

client_type_counts = pd.crosstab(
    df['SK_ID_CURR'],
    df['NAME_CLIENT_TYPE']
)

client_type_counts.columns = [
    f'PREV_APP_CLIENT_{col.upper().replace(" ","_")}_COUNT'
    for col in client_type_counts.columns
]

client_type_counts = client_type_counts.drop(
    columns=['PREV_APP_CLIENT_XNA_COUNT'],
    errors='ignore'
)

client_type_counts.head(2)
```

```python
# ============================================================
# 6. PORTFOLIO COUNTS
# ============================================================

portfolio_counts = pd.crosstab(
    df['SK_ID_CURR'],
    df['NAME_PORTFOLIO']
)

portfolio_counts.columns = [
    f'PREV_APP_PORTFOLIO_{col.upper()}_COUNT'
    for col in portfolio_counts.columns
]

portfolio_counts = portfolio_counts.drop(
    columns=['PREV_APP_PORTFOLIO_XNA_COUNT'],
    errors='ignore'
)

portfolio_counts.head(2)
```

```python
# ============================================================
# 7. PRODUCT TYPE COUNTS
# ============================================================

product_type_counts = pd.crosstab(
    df['SK_ID_CURR'],
    df['NAME_PRODUCT_TYPE']
)

product_type_counts.columns = [
    f'PREV_APP_PRODUCT_{col.upper().replace("-", "_")}_COUNT'
    for col in product_type_counts.columns
]

product_type_counts = product_type_counts.drop(
    columns=['PREV_APP_PRODUCT_XNA_COUNT'],
    errors='ignore'
)

product_type_counts.head(2)
```

```python
# ============================================================
# 9. COMBINE ALL PREVIOUS-APPLICATION FEATURES
# ============================================================

final_previous_application = (
    prev_all
    .join(status_counts, how='outer')
    .join(contract_type_counts, how='outer')
    .join(client_type_counts, how='outer')
    .join(portfolio_counts, how='outer')
    .join(product_type_counts, how='outer')
)


# Missing values after joining (NAN)
final_previous_application = (
    final_previous_application
    .fillna(0)
)

# RESET INDEX
final_previous_application = (
    final_previous_application
    .reset_index()
)

final_previous_application.head()
```

```python
# ============================================================
# 10. FINAL CHECK
# ============================================================

print("Shape:",
      final_previous_application.shape)

print(
    "Unique customers:",
    final_previous_application['SK_ID_CURR'].nunique())

print(
    "Duplicate customers:",
    final_previous_application['SK_ID_CURR'].duplicated().sum())

print("\nMissing values:")
display(
    final_previous_application
    .isna()
    .sum()
    .sort_values(ascending=False)
    .head(5)
)
```
